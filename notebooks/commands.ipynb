non-labeled dataset

```python
import os
import pandas as pd

csv = '/mnt/sda1/Repos/a-eye/Output/mri_qc/fetal/fetalqc_non-labeled-dataset/bids_csv.csv'
input_folder = '/mnt/sda1/Repos/a-eye/Output/mri_qc/fetal/fetal_qc_100'
output_folder = '/mnt/sda1/Repos/a-eye/Output/mri_qc/fetal/eye_qc_100'


# Read the CSV file
df = pd.read_csv(csv)

# Update the paths in the dataframe
df['im'] = df['im'].str.replace('/home/jaimebarranco/Desktop/non_labeled_dataset_bids/', '/mnt/sda1/Repos/a-eye/Data/SHIP_dataset/non_labeled_dataset/non_labeled_dataset_bids/')
df['mask'] = df['mask'].str.replace('/home/jaimebarranco/Desktop/non_labeled_dataset_bids/', '/mnt/sda1/Repos/a-eye/Data/SHIP_dataset/non_labeled_dataset/non_labeled_dataset_bids/')

# Get the list of filenames in the folder
filenames = [os.path.splitext(f)[0].split('_')[0] for f in os.listdir(input_folder) if os.path.isfile(os.path.join(input_folder, f))]

# Filter the dataframe
filtered_df = df[df['name'].isin(filenames)]

# Reset the index
filtered_df.reset_index(drop=True, inplace=True)

# Save the filtered dataframe to a new CSV file
filtered_df.to_csv(os.path.join(output_folder, 'bids_csv.csv'), index=False)
```
